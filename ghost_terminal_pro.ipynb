```python
# ============================================================
# 🎃 ALL-IN-ONE GHOST TERMINAL PRO (COMBINED RUNTIME CORE)
# ============================================================
!pip install -q -U google-genai gradio requests

import gradio as gr
from google import genai
from google.genai import types
from google.colab import userdata

# 1. Native Secure Credentials Handshake
API_KEY = userdata.get("GEMMA_API_KEY") or userdata.get("OPENROUTER_API_KEY")

if not API_KEY:
    raise ValueError("CRITICAL: GEMMA_API_KEY not found in your Colab Secrets panel! Please enable notebook access.")

# Directly initializing the correct unified generation client right here
ghost_client = genai.Client(api_key=API_KEY)
MODEL_NODE = "gemma-4-26b-a4b-it"

# 2. Haunted AI Processing Logic Block
def possess_my_input_text(user_input_logs):
    if not user_input_logs.strip():
        return "## 👻 Ghost Terminal: *Silence fills the dark room... Enter text to wake the machine.*"

    system_instruction = (
        "You are an ancient 'Haunted AI Terminal' possessed by a cyber-ghost.\n"
        "Analyze the user's input text or code and return a response strictly in this exact format:\n\n"
        "### 🔮 The Ghost's Whisper:\n[Provide a short, 2-sentence creepy, eerie, or gothic dark-humor commentary or code review based on their input]\n\n"
        "### 💀 Hidden Digital Curse:\n- [One funny, spooky tech curse, e.g., Your loop will run for eternity, or A phantom indent error will hunt your repository]\n\n"
        "### 🕯️ Día de los Muertos Challenge:\n- [Give one instruction honoring history, e.g., Read one old code archive, or Light a candle and turn off your internet for 10 minutes]"
    )

    try:
        # Firing content generation loops seamlessly over the local client object
        response = ghost_client.models.generate_content(
            model=MODEL_NODE,
            contents=f"USER INPUT DATA LOGS:\n{user_input_logs}",
            config=types.GenerateContentConfig(
                system_instruction=system_instruction,
                temperature=0.2
            )
        )
        return response.text.strip()

    except Exception as e:
        return (
            f"## ⚠️ Portal Request Failed\n\n"
            f"Google AI Studio engine tracking timeout or validation failure.\n\n"
            f"**Technical Details:** `{str(e)}`"
        )

# 3. Designing Visual presentation UI Canvas
with gr.Blocks(theme=gr.themes.Soft(), title="Ghost Terminal Pro") as demo_canvas:
    gr.Markdown("# 🎃 GHOST TERMINAL PRO\n### 🎓 Week 4 Challenge: Open-Source AI 'Ghost in the Machine' Terminal Framework")
    gr.Markdown("An interactive haunted dashboard. Type your daily logs, thoughts, or code snippets, and let the open-weight model possess your execution pipeline.")

    with gr.Row():
        with gr.Column():
            input_box = gr.Textbox(label="📥 Enter Your Text / Code Log to Process", placeholder="Type something innocent here, e.g., print('Hello World') or My system is running fast...", lines=6)
            submit_btn = gr.Button("Submit to the Ghost Machine", variant="primary")
        with gr.Column():
            output_panel = gr.Markdown(value="### 🕸️ Eerie Transmission Records\nYour possessed, structured text translations will render here dynamically.")

    submit_btn.click(fn=possess_my_input_text, inputs=input_box, outputs=output_panel)

# Deploying and revealing live share tunnels
demo_canvas.launch(share=True, debug=True)
```
